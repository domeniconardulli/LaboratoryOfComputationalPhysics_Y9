```python
stringa = ""

for i in range(1,101):
    if i%3 ==0 and i%5 == 0:
        stringa += "MickeyMouse "
    elif i%3 == 0:
        stringa += "Mickey "
    elif i%5 == 0:
        stringa += "Mouse "
    else:
        stringa += "%d "%i

print(stringa)
    
    
    
```

ESERCIZIO 1-B

```python
lista = stringa.split(" ")


for i in range(0,len(lista)):
    if lista[i] == "Mickey":
        lista[i] = "Donald"
    if lista[i] == "Mouse":
        lista[i] = "Duck"
    else:
        continue

print(tuple(lista))
```

ESERCIZIO 3 (DISTANZA)

```python
from math import sqrt

u = (3,0)
v = (0,4)

d = sqrt((u[0]-u[1])**2 + (v[0]-v[1])**2)

print(d)
```
