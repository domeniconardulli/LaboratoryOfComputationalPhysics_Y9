esercizio 1 (write the following as list comprehensions)

```python
ans = []
for i in range(3):
    for j in range(4):
        ans.append((i, j))
print (ans)

# 2
ans = map(lambda x: x*x, filter(lambda x: x%2 == 0, range(5)))
print (list(ans))

#risposte:
risp = [(i,j) for i in range(3) for j in range(4)]
print(risp)

risp2 = [x**2 for x in range(5) if x%2 == 0]
print(risp2)
```

esercizio 4 Write the factorial function so that it a) does and b) does not use recursion.

```python
def fattricors(n):
    if n == 0 or n ==1:
        return 1
    elif n > 1:
        return n*fattricors(n-1)
    
print(fattricors(4))

def fattiter(n):
    y = n-1
    f = n
    while y > 0:
        f = f*y
        y = y-1
    return f

print(fattiter(5))
```
