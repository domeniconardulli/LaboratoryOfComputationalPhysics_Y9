```python
#programma che converte decimale in binario ecc...
import random

def binario(x):
    i = ''
    while x > 0:
        y = int(x)
        x = x//2
        i += str(y%2)
    j = ''
    for k in range(len(i)):
        j += i[-k-1]
    return j
#______________________________________________________________________
y = random.randint(0,10000)

print(y)

if str(bin(y)) == '0b' + binario(y):
    print('funziona')
    
#____________________________________________________________________

def esadecimale(x):
    i = ''
    while x > 0:
        y = int(x)
        x = x//16
        if y%16 <= 9:
            i += str(y%16)
        if y%16 == 10:
            i += 'a'
        if y%16 == 11:
            i += 'b'
        if y%16 == 12:
            i += 'c'
        if y%16 == 13:
            i += 'd'
        if y%16 == 14:
            i += 'e'
        if y%16 == 15:
            i += 'f'
    j = ''
    for k in range(len(i)):
        j += i[-k-1]
    return j


z = random.randint(0,10000)

print(hex(z))
print('0x' + esadecimale(z))

#________________________________________________________________    
def decimale(x):
    y = 0
    z = str(x)
    j = ''
    for k in range(len(z)):
        j += z[-k-1]
    for i in range(len(j)):
        y += int(str(j)[i])*(2**i)
    return y

print(decimale(10010))



        
        
        
    
```
