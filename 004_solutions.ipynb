```python
import numpy as np
```

```python
#ex. 1
m = np.arange(12).reshape((3,4))
print(m)
print(np.mean(m))
```

```python
#ex. 2

u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
v = v[:,None] #questo traspone un vettore, con broadcasting:al posto di None puoi mettere np.newaxis
z = u*v
print(v, '\n',z)
```

```python
#ex. 3

import numpy.random as npr

y = npr.rand(10,6)
mask = (y > 0.1)
print(mask)
x = y*mask
print(x)
```

```python
#ex. 4

import matplotlib.pyplot as plt

A = np.linspace(0, 2*np.pi, 100)
print(A)

print(A[[x*10 for x in range(0,10)]])

print(A[::-1]) #questo inverte l'array: vedi perche'

mask1 = (np.absolute(np.sin(A) - np.cos(A)) < 0.1)

W = A[mask1].copy()

print(W)

plt.plot(A, np.cos(A))
plt.plot(A, np.sin(A)) #CERCA DI CAPIRE PERCHE' RIEMPIE
plt.show
```

```python
#ex. 5

X = np.arange(1,11)
Y = X[:,None]
print(X*Y)
Z = X*Y

print(np.sum(Z*np.eye(10))) #traccia

Z2 = Z[[-x for x in range(1,11)],:]
print(Z2)

Z3 = Z2*np.eye(10)

mask2 = Z3 > 0

print(Z3[mask2])

Z2 = Z[[-x for x in range(0,10)],:]

print(Z2)

Z3 = Z2*np.eye(10)

mask2 = Z3 > 0

print(Z3[mask2])
```
